```python
#MakeMore
#这一章的主要目的是实现一个姓名生成模型
#通过一个巨大的姓名数据集训练出一个可以生成类似名字的模型
#这里其实实现的就是一个bigram模型
```

```python
#导入姓名数据集
words=open("names.txt",'r').read().splitlines()
```

```python
words[:10]
```

```python
#数据集长度
len(words)
```

```python
#最短的名字
min(len(w)for w in words)
```

```python
#最长的名字
max(len(w)for w in words)
```

```python
#我们期望建立一个双字符形状的模型
#也就是通过前一个字符预测下一个字符
#那我们第一件事就是要把前后两个字符两两配对
#现在展示配对第一个名字的代码
```

```python
for w in words[:1]:
    for ch1,ch2 in zip(w,w[1:]):
        print(ch1,ch2)
```

```python
#这里可以看到我们把第一个名字emma两两配对成了em,mm,ma
```

```python
#但是有一个问题：模型不仅要预测下一个字母是什么，还要能判断开头字母会是什么以及名字会在什么时候结束
#所以我们可以在每个名字的开头结尾加上标识
```

```python
for w in words[:1]:
    chs=['<S>']+list(w)+['<E>']
    for ch1,ch2 in zip(chs,chs[1:]):
        print(ch1,ch2)
```

```python
#可以看到现在就有了开头<S>和结尾<E>的标识 
#现在可以尝试三个名字
```

```python
for w in words[:3]:
    chs=['<S>']+list(w)+['<E>']
    for ch1,ch2 in zip(chs,chs[1:]):
        print(ch1,ch2)
```

```python
#也成功显示出开头结尾
```

```python
#我们需要实现一个预测模型，那自然就需要统计每种组合出现的次数
#所以建立一个字典来存储组合的数量
```

```python
b={}
for w in words[:3]:
    chs=['<S>']+list(w)+['<E>']
    for ch1,ch2 in zip(chs,chs[1:]):
        bigram = (ch1,ch2)#这里的bigram是一个元组，元组可以作为字典的键
        b[bigram]=b.get(bigram,0)+1
        #这里的b.get()的意思是如果存在bigram就返回对应的值，不存在返回0
    #所以最后的b代表了每一种组合的计数
b
```

```python
b={}
for w in words:#现在统计全部字符
    chs=['<S>']+list(w)+['<E>']
    for ch1,ch2 in zip(chs,chs[1:]):
        bigram = (ch1,ch2)
        b[bigram]=b.get(bigram,0)+1

b
```

```python
b.items()#返回一个items对象，把原字典里的每一对包装成一个（键，值）的元组
```

```python
sorted(b.items(),key = lambda kv:kv[1])
#排序语法，意思是把items里的每一个元素按照一定的顺序排列
#参考标准为lambda
#lambda 参数:返回值
#所以这里说的是按照每一个元组的第二个值，也就是次数来排序
```

```python
#现在我们来思考一下之前我们想建立的计数表
#总共有26个字母（没有大写，首字母已经有<S>标识）
#加上<S><E>，总共有28个字符
#所以一个28*28的张量就可以表示所有的频数
```

```python
N = torch.zeros((28,28),dtype=torch.int32)
```

```python
#接下来我们需要解决下一个问题：
#tensor的向量值只能是数字而不能是字母，也就是说不存在N[a][b]这样的东西
#所以我们需要建立字符表
#也就是说我们需要建立字符->数字和数字->字符的映射
#首先来建立字符表
```

```python
#这里的逻辑是：
#''.join(words)将words整合成了一个巨大的string
#set去重，保留下所有的字符种类
#list转换成，可以排序
#sorted进行排序，得到名字中出现的所有字母的字母表
chars = sorted(list(set(''.join(words))))
chars
```

```python
#接下来就可以建立一一对应的表了
#也就是我们可以建立两个字典
#{字符:数字}&{数字:字符}
#两个表分别是stoi和itos，也就是string to int和int to string
```

```python
#这里的enumerate是一个迭代器
#每一次会返回一个(数字，字符)形状的元组
stoi={s:i for i,s in enumerate(chars)}
#添加上<S><E>
stoi['<S>']=26
stoi['<E>']=27
stoi
```

```python
#顺便建立下itos
itos={i:s for i,s in enumerate(chars)}
itos[26]='<S>'
itos[27]='<E>'
itos
```
