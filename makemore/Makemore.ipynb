```python
#现在我们来思考一下之前我们想建立的计数表
#总共有26个字母（没有大写，首字母已经有<S>标识）
#加上<S><E>，总共有28个字符
#所以一个28*28的张量就可以表示所有的频数
```

```python
N = torch.zeros((28,28),dtype=torch.int32)
```

```python
#接下来我们需要解决下一个问题：
#tensor的向量值只能是数字而不能是字母，也就是说不存在N[a][b]这样的东西
#所以我们需要建立字符表
#也就是说我们需要建立字符->数字和数字->字符的映射
#首先来建立字符表
```

```python
#这里的逻辑是：
#''.join(words)将words整合成了一个巨大的string
#set去重，保留下所有的字符种类
#list转换成，可以排序
#sorted进行排序，得到名字中出现的所有字母的字母表
chars = sorted(list(set(''.join(words))))
chars
```

```python
#接下来就可以建立一一对应的表了
#也就是我们可以建立两个字典
#{字符:数字}&{数字:字符}
#两个表分别是stoi和itos，也就是string to int和int to string
```

```python
#这里的enumerate是一个迭代器
#每一次会返回一个(数字，字符)形状的元组
stoi={s:i for i,s in enumerate(chars)}
#添加上<S><E>
stoi['<S>']=26
stoi['<E>']=27
stoi
```

```python
#顺便建立下itos
itos={i:s for i,s in enumerate(chars)}
itos[26]='<S>'
itos[27]='<E>'
itos
```

```python
#接下来有了字符数字对照表，我们可以开始着手操作N了
for w in words:
    chs=['<S>']+list(w)+['<E>']
    for ch1,ch2 in zip(chs,chs[1:]):
        n1=stoi[ch1]
        n2=stoi[ch2]#将字符转换成数字
        N[n1,n2]+=1
```

```python
#好的，现在我们已经完整操作完之前的那个数据集了
#下一步就是着手设计模型了
#这里andrej首先采用的是bigram模型，也就是的单字节模型，只用一个字符作为参考去生成下一个字符

#bigram的逻辑是这样的：根据前一个字符去推测下一个字符，而不参考前文
#而根据统计学的话，这样的方式最好的使用频率代替概率
#比如说统计出来a后面60%接b，40%接c，那这个就可以看作是概率

#所以说：我们之前统计出的每一行的频率
#也就可以直接转化为相应的bigram概率

#注意！
#这里的bigram看着就是一个表格而不是训练出的模型
#其实他就是一个不需要任何训练，直接把频率转化为概率的模型
#每一个格子其实对应的是：当前模型对于已经给出的这个字母（行参数），所给出的下一个字母（列参数）的可行度参数
#比如说对于[b][c]格子，就是说模型认为现在的字母是b的话下一个字母是c的可能性有多大
#所以我们也可以理解为：之前的建表的过程其实就已经在训练模型了

#因此这个bigram模型可以这样理解：每一行频率（概率）代表了一个27个面的骰子
#总共有27个骰子
#比如骰子a，就代表了a之后有27中可能
#但显然这个骰子不是均匀的而是有轻重之分的
#所以说我们需要实现一颗可以自由操控每一面概率的骰子
#pytorch实现了一个函数叫做多项式采样函数，multinomial()
#这个函数可以帮我们实现这颗骰子
#现在我们来展示一下
```

```python
#我们现在使用一个随机数生成器，生成一个有3个面的不均匀骰子来展示一下
g=torch.Generator().manual_seed(2147483647)#随机数生成器，后面的是种子，固定种子可以确保随机过程中的结果是一样的
p1=torch.rand(3,generator=g)#这里意思是根据种子生成三个0～1的随机数
p1=p1/p1.sum()#归一化处理
p1#此时，p1就是一个有三个面的骰子
```

```python
torch.multinomial(p1,num_samples=100,replacement=True,generator=g)
#这就是多项式采样函数
#第一个参数tensor代表了这个骰子长什么样，我们之前已经有一个p作为参数骰子了
#其实第一个参数会自动归一化，但是我们手动实现了一下
#第二个参数num_samples表示扔多少次这个骰子
#第三个参数replacement是不是有放回的抽取，这里显然是有放回的，也就是说前面抽出什么不影响下一次抽
#第四个参数generator是随机数生成器
```

```python
#可以看到根据概率我们投了这个骰子十次，且基本符合概率分布
#那这样的话显然我们可以把这个函数作用到之前的N上
```

```python
g=torch.Generator().manual_seed(2147483647)
ix=torch.multinomial(p,num_samples=1,replacement=True,generator=g).item()#这里只取样一次，同时使用item()将tensor转化为int，就实现了有概率差别的27面骰子
ix#这里由于pytorch的更新，相同种子的结果会和andrej的结果不一样，andrej的结果是13
```

```python
#得到了数字结果就可以转换成对应的字符
itos[ix]
```

```python
#所以我们用刚刚的种子和生成器根据第一个是'.'抽取到了下一个是'c'
#那对于文本生成，所以下一步我们的方案就是根据这个'c'选取下一个可能的值
#并且以此类推
#所以我们现在应该可以开始写循环了
```

```python
g=torch.Generator().manual_seed(2147483647)

out=[]
ix=0
while True:
    p=N[ix].float()
    p=p/p.sum()#取概率
    ix=torch.multinomial(p,num_samples=1,replacement=True,generator=g).item()
    out.append(itos[ix])
    if ix==0:
        break
print(''.join(out))
```

```python
#这个循环给我们生成了一个看着像模像样的名字cexze
#让我们多生成几个
#这里我们需要使用到一些并行化技巧了
#之前的每一次循环我们都要进行p=N[ix].float() p=p/p.sum()，非常低效
#所以我们建立一个大的表P，每一个格子直接代表了在这一行的归一化结果
```

```python
g=torch.Generator().manual_seed(2147483647)

P=N.float()#浮点化
P.sum(1,keepdim=True)
#这里有一个非常关键的函数sum
#现在我们希望用P/P.sum来求概率
#但是这个P.sum应该是每一行的求和值
#这里的两个参数
#第一个1代表着按照第1个维度去求和
#换句话说是压缩那个分量
#也就是说按照把27个列压缩成一个列
#第二个参数非常关键
#第二个参数代表的意思是是否要保留dimension维度
#默认的是false，也就是说默认会压缩维度的
#默认的话原来是一个二维的表格，压缩完了就变成了一个一维的向量，跟个列表差不多
#但是去赋值为True的话得到的依旧是一个二维的表格，只不过变成了从27*27变成了27*1
```

```python
#可以看到按照每一行去求和了，接下来可以继续操作了
P/=P.sum(1,keepdim=True)#/=原地除法可以节省内存
#这里可以用除法的原因是pytorch对于除法的规定是：维度一样且对应的每个维度要么值一样要么有一个是1
#这里的两个分别是
#27,27
#27,1
#满足条件所以可以直接运用除法
#1就会将它复制展开成27个一模一样的去挨个做除法
P
```

```python
#这里的keepdim非常重要
#pytorch的除法存在广播机制
#广播机制的原理是向右对齐维度之后，每一对一一对应的维度要么相等要么有一个为1要么有一个不存在
#我们来看看刚刚P.sum(1,keepdim=True)的结果
#再看看如果没有设置这个keepdim会是什么样的
print(P.sum(1,keepdim=True).shape)
print(P.sum(1).shape)#keepdim缺省为False
```

```python
#可以看到二者在内容上没有区别，但是在形状上有区别
#一个是二维向量，一个是一维向量
#比如同一个'c'
#对于前者就是[3][0]
#对于后者就是单纯的[3]
```

```python
#这个时候我们放到除法中去看
#根据广播机制，二维向量会这样对应
#27 27 
#27  1
#这个时候进行除法，会将第二个维度复制展开
#于是第二个27*1变成了27个相等的27*1组成了27*27
#而第二种情况会是这样的：
#27 27
#   27（向右对齐）
#这个时候，pytorch会自动补全没有的维度，于是乎
#27 27
# 1 27
#显然就不对了，因为这里会把它看作一行数据然后平铺
#也就是每一列的除数是一样的了，但是我们想要的是每一行是一样的
#所以其实我们可以发现：当维度被压缩时候，需要实时关注广播机制的变化，最好的办法是操作维数相同的向量
```

```python
g=torch.Generator().manual_seed(2147483647)

for i in range(20):#生成20个名字
    out=[]
    ix=0
    while True:
        p=P[ix]
        ix=torch.multinomial(p,num_samples=1,replacement=True,generator=g).item()
        out.append(itos[ix])
        if ix==0:
            break
    print(''.join(out))
```

```python
#成功使用P生成20个名字
#可以看到有写名字生成的还可以，比如aisan,teda
#但是大部分名字都是一团糟
#所以我们可以得到现在的bigram的模型确实是在生成名字，但是效果不好
#本质是bigram模型实在太单一，完全没有上下文机制 
#所以接下来我们就要优化我们的模型
#在优化之前，首先我们要学会如何去评估一个模型的性能如何
#还记得在micrograd里学过的loss吗？那里用到了最大似然估计
#首先要理解最大似然估计是什么东西
```

```python
#这个东西拆开来最核心的部分就是似然
#刚刚说我们的那个P看着是一个表格其实上是一个模型
#模型的作用在于对之前上万个名字的规律进行解释
#这个东西就叫做似然
#但是呢这个解释有很多
#也就是说这样的27*27的表格有成千上万过
#于是乎我们就需要找到那个最能解释我们之前的姓名数据集的模型
#我们刚刚用频率生成概率的过程就是找到那个最好的模型
#而这个过程就叫最大似然估计
#去估计到那个解释的最好的模型
#但是模型再好也有误差
#而这个误差就是它给出的正确答案的置信度和1之间的差距
#这个求出来的东西就是loss
```

```python
#也就是说我们用的是（1-模型对于正确答案给出的预估）作为评估模型的参数

#我们之前的bigram的每一个格子代表了对于下一个字母的预测值
#所以由此我们可以算出这个bigram的loss

#先来采用之前的算法来提取真实姓名并看一下模型给出的预估概率是多少
for w in words[:3]:
    chs=['.']+list(w)+['.']
    for ch1,ch2 in zip(chs,chs[1:]):
        n1=stoi[ch1]
        n2=stoi[ch2]
        prob=P[n1,n2]
        print(f'{ch1}{ch2}={prob:.4f}')
```

```python
#这样看的话有很多的loss，我们怎么把他们整合成一个loss呢？

#我们可以这样看：每一个插值都是一次独立的投骰子
#所以根据乘法原理我们可以把他们乘起来

#但是有一个问题就是这些数字都太小了
#乘起来会变成一个非常小的值

#所以我们可以加工一下我们的数据
#给他们套上一个ln来放大数据
#同时ln还能把乘法转换成加法
#所以这里使用ln还可以简化将来的乘积运算
#并且我们可以直接用加法得到最后的乘积

log_likelihood=0
for w in words[:3]:
    chs=['.']+list(w)+['.']
    for ch1,ch2 in zip(chs,chs[1:]):
        n1=stoi[ch1]
        n2=stoi[ch2]
        prob=P[n1,n2]
        logprob=torch.log(prob)#这里的底数是e，也就是说这里是ln(prob)
        log_likelihood+=logprob
        
        print(f'{ch1}{ch2}={prob:.4f},{logprob:.4f}')
print(f'{log_likelihood=}')
```

```python
#这里可以看到我们得到了一个值：-38.7856
#但是我们希望得到一个正数的参考值loss
#且模型越符合正解，这个loss越接近0
#于是我们可以直接把likelihood取反
#这就得到了负对数似然
nll=-log_likelihood
print(f'{nll=}')
```

```python
#同时呢，我们评价模型肯定和统计数量无关，但是我们刚刚直接累加
#所以一般情况下我们可以再取个平均

log_likelihood=0
n=0
for w in words[:3]:
    chs=['.']+list(w)+['.']
    for ch1,ch2 in zip(chs,chs[1:]):
        n1=stoi[ch1]
        n2=stoi[ch2]
        prob=P[n1,n2]
        logprob=torch.log(prob)
        log_likelihood+=logprob
        
        nll=-log_likelihood
        n+=1
        print(f'{ch1}{ch2}={prob:.4f},{logprob:.4f}')
print(f'{log_likelihood=}')

print(f'{nll=}')
print(f'average_nll={nll/n}')
```

```python
#这样得到的结果就和我们取多少样本检测无关了
#然后按照这个思路我们再来检测一下这个模型整体的nll
```
